```python
# Criar nova coluna season_nome

df_hour['season_nome'] = df_hour['season'].map({
    1: 'Primavera',
    2: 'Verão',
    3: 'Outono',
    4: 'Inverno'})

display(df_hour)

```

```python
# A coluna yr usa códigos para codificar o ano

df_hour['yr'].value_counts().sort_index()

```

```python
# Criar nova coluna ano

df_hour['ano'] = df_hour['yr'].map({
    0: 2011,
    1: 2012
})

display(df_hour)
```

```python

# A coluna mnth usa códigos para codificar o mês

df_hour['mes_nome'] = df_hour['mnth'].map({
    1: 'Janeiro',
    2: 'Fevereiro',
    3: 'Março',
    4: 'Abril',
    5: 'Maio',
    6: 'Junho',
    7: 'Julho',
    8: 'Agosto',
    9: 'Setembro',
    10: 'Outubro',
    11: 'Novembro',
    12: 'Dezembro'
})

display(df_hour)
```

```python
# a variável workingday considera se é dia útil ou não útil

df_hour['tipo_dia'] = df_hour['workingday'].map({
    0: 'Não útil',
    1: 'Dia útil'
})

display(df_hour)
```

```python
# condificação das condiçôes climaticas

df_hour['weathersit'].value_counts().sort_index()
```

```python
# Criar a coluna clima

df_hour['clima'] = df_hour['weathersit'].map({
    1: 'Poucas nuvens/Céu limpo',
    2: 'Nublado/Névoa',
    3: 'Chuva leve/Neve',
    4: 'Chuva forte'
})

display(df_hour)
```

```python

# Criar a coluna dia da semana

df_hour['dia_semana'] = df_hour['weekday'].map({
    0: 'Domingo',
    1: 'Segunda-feira',
    2: 'Terça-feira',
    3: 'Quarta-feira',
    4: 'Quinta-feira',
    5: 'Sexta-feira',
    6: 'Sábado'
})

display(df_hour)
```

```python
# Pergunta 1: Qual foi a quantidade média de bicicletas alugadas por hora em cada ano?

media_por_ano = (
    df_hour.groupby('ano')['cnt']
    .mean()
    .round(2)
)

display(media_por_ano)

diferenca = media_por_ano.loc[2012] - media_por_ano.loc[2011]

display (diferenca)

resposta1 = "Resposta:Em 2011, a quantidade média de bicicletas alugadas foi de 143,79 por hora, enquanto em 2012 foi de 234,67 por hora. Portanto, houve um aumento de 90,88 bicicletas na média de aluguéis por hora entre os dois anos."

print(resposta1)
```

```python
# Pergunta 2: Em qual horário do dia ocorreu a maior média de bicicletas alugadas?

media_por_hora = (
    df_hour.groupby('hr')['cnt']
    .mean()
    .round(2)
)

display (media_por_hora)

horario_maior_media = media_por_hora.idxmax()
maior_media = media_por_hora.max()

print("Horário com maior média:", horario_maior_media, "horas")
print("Média de bicicletas alugadas:", maior_media)

resposta2 = "Resposta: O horário com a maior média de bicicletas alugadas foi 17h, com aproximadamente 461,45 bicicletas por hora. Também foram observadas médias elevadas às 18h e às 8h. Esses resultados mostram uma maior concentração de aluguéis em determinados horários do dia."

print(resposta2)
```

```python
# Pergunta 3: Qual mês apresentou a maior média de bicicletas alugadas por hora?

media_por_mes = (
    df_hour.groupby('mes_nome')['cnt']
    .mean()
    .round(2)
    .sort_values(ascending=False)
)
display (media_por_mes)


resposta3 = "Resposta: O mês que apresentou a maior média de bicicletas alugadas por hora foi setembro, com aproximadamente 240,77 bicicletas por hora."

print(resposta3)
```

```python
# Pergunta 4: A média de bicicletas alugadas é maior nos dias úteis ou nos dias não úteis?

media_tipo_dia = (
    df_hour.groupby('tipo_dia')['cnt']
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

display(media_tipo_dia)



resposta4 = "A média de bicicletas alugadas foi maior nos dias úteis, com aproximadamente 193,21 bicicletas por hora"
print(resposta4)
```

```python
# Pergunta 5: Em quais horários do dia os usuários ocasionais e registrados apresentam a maior quantidade total de aluguéis? Informe o horario de pico para cada classe de usuário

alugueis_por_hora_usuario = (
    df_hour.groupby('hr')[['casual', 'registered']]
    .sum()
    
)

print(alugueis_por_hora_usuario)

hora_pico_casual = alugueis_por_hora_usuario['casual'].idxmax()
hora_pico_registrado = alugueis_por_hora_usuario['registered'].idxmax()

print("Horário de pico dos usuários casuais:", hora_pico_casual, "horas")
print("Horário de pico dos usuários registrados:", hora_pico_registrado, "horas")

resposta5 = "Resposta: Os usuários casuais tem horario de pico as 14:00 h e os usuários registrados as 17:00 h."

print(resposta5)

```

```python
# Pergunta 6: Como se distribui a quantidade total de aluguéis entre usuários casuais e registrados em dias úteis e não úteis?

alugueis_tipo_dia = (
    df_hour.groupby('tipo_dia')[['casual', 'registered']]
    .sum()
)

alugueis_tipo_dia['total'] = (
    alugueis_tipo_dia['casual'] +
    alugueis_tipo_dia['registered']
)

alugueis_tipo_dia['% casual'] = (
    alugueis_tipo_dia['casual'] /
    alugueis_tipo_dia['total'] * 100
).round(2)

alugueis_tipo_dia['% registered'] = (
    alugueis_tipo_dia['registered'] /
    alugueis_tipo_dia['total'] * 100
).round(2)

alugueis_tipo_dia['% total'] = (
    alugueis_tipo_dia['% casual'] +
    alugueis_tipo_dia['% registered']
)

display(alugueis_tipo_dia)

resposta6 = "Resposta: Nos dias úteis, os usuários registrados representam 86,77% dos aluguéis, enquanto os casuais correspondem a apenas 13,23%. Já nos dias não úteis, a participação dos usuários casuais aumenta para 31,66%, enquanto a dos registrados cai para 68,34%."

print(resposta6)
```

```python
# Pergunta 7: Qual é o impacto das condições climáticas adversas no total de aluguéis de bicicletas?


analise_clima = (
    df_hour.groupby('clima')
    .agg(
        quantidade_registros=('cnt', 'count'),
        total_alugueis=('cnt', 'sum')
    )
)

analise_clima['alugueis_por_registro'] = (
    analise_clima['total_alugueis'] /
    analise_clima['quantidade_registros']
).round(2)


tempo_bom = analise_clima.loc[
    'Poucas nuvens/Céu limpo',
    'alugueis_por_registro'
]

print(tempo_bom)


analise_clima['reducao_%'] = (
    (tempo_bom - analise_clima['alugueis_por_registro'])
    / tempo_bom * 100
).round(2)

analise_clima = analise_clima.sort_values(
    by='alugueis_por_registro',
    ascending=False
)

display(analise_clima)

resposta7 = "Resposta: Em relação às condições de poucas nuvens/céu limpo, houve redução de 14,50% nos aluguéis por registro em condições de tempo nublado/névoa, 45,54% em chuva leve/neve e 63,72% em chuva forte. Entretanto, a categoria chuva forte possui apenas 3 registros, portanto seu resultado deve ser interpretado com cautela."

print(resposta7)
```

```python
# Pergunta 8: Como os aluguéis de bicicletas se distribuem entre diferentes faixas de temperatura?

df_hour['temp'].describe()

```

```python
# Pergunta 8: Como os aluguéis de bicicletas se distribuem entre diferentes faixas de temperatura?

df_hour['faixa_temperatura'] = pd.cut(
    df_hour['temp'],
    bins=[0, 0.34, 0.50, 0.66, 1.0],
    labels=[
        'Baixa',
        'Moderada-baixa',
        'Moderada-alta',
        'Alta'
    ],
    include_lowest=True
)

df_hour[['temp', 'faixa_temperatura']].head(10)
```

```python
# Pergunta 8: Como os aluguéis de bicicletas se distribuem entre diferentes faixas de temperatura?

alugueis_por_temperatura = (
    df_hour.groupby(
        'faixa_temperatura',
        observed=True
    )['cnt']
    .sum()
)

display(alugueis_por_temperatura)


percentual_temperatura = (
    alugueis_por_temperatura /
    alugueis_por_temperatura.sum() * 100
).round(2)

display(percentual_temperatura)


analise_temperatura = pd.DataFrame({
    'total_alugueis': alugueis_por_temperatura,
    'percentual_%': percentual_temperatura
})

analise_temperatura = analise_temperatura.sort_values(
    by='total_alugueis',
    ascending=False
)

display(analise_temperatura)

resposta8 = "Resposta: A análise mostra que a maior concentração de aluguéis ocorreu na faixa de temperatura classificada como Alta, com 1.108.280 aluguéis, correspondendo a 33,66% do total. Em seguida, aparece a faixa Moderada-alta, com 1.027.577 aluguéis (31,21%). A faixa Moderada-baixa concentrou 691.027 aluguéis (20,99%), enquanto a faixa de temperatura Baixa apresentou a menor participação, com 465.795 aluguéis (14,15%)." \

print(resposta8)
```

```python
# Pergunta 9 — Quais foram os 10 dias com maior número de aluguéis e quais características esses dias apresentam?


alugueis_por_dia = (
    df_hour.groupby('dteday')
    .agg(
        total_alugueis=('cnt', 'sum'),
        ano=('ano', 'first'),
        mes=('mes_nome', 'first'),
        dia_semana=('dia_semana', 'first'),
        tipo_dia=('tipo_dia', 'first')
    )
    .reset_index()
)

top_10_dias = alugueis_por_dia.nlargest(
    10,
    'total_alugueis'
)

display(top_10_dias)


print("Quantidade por ano:")
print(top_10_dias['ano'].value_counts())

print("\nQuantidade por mês:")
print(top_10_dias['mes'].value_counts())

print("\nQuantidade por dia da semana:")
print(top_10_dias['dia_semana'].value_counts())

print("\nQuantidade por tipo de dia:")
print(top_10_dias['tipo_dia'].value_counts())

resposta9 = "Resposta: Os 10 dias com maior número de aluguéis ocorreram todos em 2012. O maior volume foi registrado em 15 de setembro de 2012, com 8.714 aluguéis. Setembro destacou-se entre os meses, aparecendo em 5 dos 10 dias de maior movimento. Quanto ao tipo de dia, 6 dos 10 registros ocorreram em dias não úteis e 4 em dias úteis. Entre os dias da semana, o sábado aparece 5 vezes, seguido pela sexta-feira, com 3 ocorrências. Os resultados mostram, portanto, uma concentração dos maiores volumes principalmente em 2012, no mês de setembro e aos sábados." \

print(resposta9)
```

```python
# Pergunta 10: Como evoluiu o total mensal de aluguéis de bicicletas entre 2011 e 2012?

alugueis_mensais = (
    df_hour.groupby(['ano', 'mnth', 'mes_nome'])['cnt']
    .sum()
    .reset_index()
)

display(alugueis_mensais)

alugueis_grafico = alugueis_mensais.pivot(
    index='mnth',
    columns='ano',
    values='cnt'
)

display(alugueis_grafico)

import matplotlib.pyplot as plt

alugueis_grafico.plot(
    kind='line',
    marker='o',
    figsize=(10, 6)
)

plt.title('Evolução mensal dos aluguéis de bicicletas — 2011 x 2012')
plt.xlabel('Mês')
plt.ylabel('Total de aluguéis')
plt.xticks(
    range(1, 13),
    ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun',
     'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
)
plt.legend(title='Ano')
plt.grid(alpha=0.3)
plt.show()

resposta10 = """Resposta: O gráfico mostra que o total mensal de aluguéis foi maior em 2012 do que em 2011 em todos os meses do ano. Em ambos os anos, observa-se um crescimento dos aluguéis nos primeiros meses, seguido por valores mais elevados durante os meses intermediários do ano e uma redução nos últimos meses.

Em 2011, o total de aluguéis cresce principalmente entre janeiro e junho, atingindo seu maior nível aproximadamente em junho, e passa a apresentar redução nos meses seguintes. Em 2012, o crescimento se mantém por um período mais longo, com o maior volume ocorrendo em setembro, seguido por uma queda mais acentuada a partir de outubro.

A comparação entre os dois anos evidencia, portanto, um aumento expressivo na utilização do sistema de compartilhamento de bicicletas em 2012, além de indicar uma variação do volume de aluguéis ao longo dos meses do ano."""

print(resposta10)
```
